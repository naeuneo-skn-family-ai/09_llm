# Self-query Retriever로 자연어 필터 생성

Self-query Retriever는 사용자의 자연어 질문에서 **검색어와 metadata 조건을 스스로 분리하는 Retriever**이다. 여기서 Self-query는 모델을 다시 학습한다는 뜻이 아니라, LLM이 검색용 구조를 만든다는 뜻이다.

### 자연어 질문에서 분리하는 세 요소

사용자가 `김민수가 작성한 검색 관련 문서 2개만 찾아줘`라고 질문하면 다음 세 가지 검색 조건이 필요하다.

- 검색어: `검색 관련`이다.
- metadata 조건: `author == 김민수`이다.
- 반환 개수: `2개`이다.

Self-query Retriever는 LLM으로 이 조건들을 자동 추출하고 Vector Store가 사용할 검색 구조로 변환한다. 이 과정에서는 **자연어 질문이 실제 검색 조건으로 바뀌는 과정**을 확인한다. 최종 반환값은 답변이 아니라 `list[Document]`이다.

### 세 검색 방식의 차이

- 일반 Dense Retriever: 질문 전체를 embedding한 뒤 의미가 가까운 문서를 찾는다. metadata 조건을 자동으로 만들지 않는다.
- Metadata Filtering: 개발자가 `{'category': {'$in': ['AI']}}` 같은 filter 딕셔너리를 직접 작성한다.
- Self-query Retriever: 사용자의 자연어 조건을 LLM이 `StructuredQuery`로 바꾸고 filter를 자동 생성한다.

### Self-query 처리 흐름

`자연어 질문 → query_constructor → StructuredQuery → translator → Pinecone 검색 → list[Document]`

- `query_constructor`: 자연어 질문을 검색어·filter·limit으로 분리한다.
- `StructuredQuery`: 분리한 검색 조건을 담는 LangChain 객체이다.
- `translator`: LangChain의 공통 filter 표현을 Pinecone 문법으로 변환한다.
- Pinecone: 변환된 검색어와 filter를 적용해 문서를 반환한다.

먼저 `04_metadata_filtering.ipynb`의 upsert까지 실행해 `adv-rag-meta`에 본문과 `author`, `category` metadata를 저장해야 한다. 이 노트북은 같은 인덱스를 검색한다.

## Self-query Retriever 구성 1: Pinecone filter translator

Self-query Retriever는 자연어를 Pinecone에 바로 전달하지 않고 다음 세 단계를 연결한다.

1. `query_constructor`: 자연어를 LangChain의 `StructuredQuery`로 변환한다.
2. `translator`: StructuredQuery의 조건을 Pinecone filter 딕셔너리로 변환한다.
3. `Vector Store`: 의미 검색어와 filter를 이용해 `list[Document]`를 반환한다.

이 코드셀은 두 번째 단계인 `PineconeFilterTranslator`를 정의한다. metadata 스키마와 최종 Retriever 객체는 뒤의 연결 셀에서 만든다.

### translator가 필요한 이유

LangChain과 Pinecone은 같은 검색 조건을 서로 다른 형식으로 표현한다. translator는 LangChain 형식을 Pinecone이 읽을 수 있는 형식으로 바꾸는 중간 변환기이다.

```text
자연어: 김민수가 작성한 문서를 찾아줘
→ StructuredQuery: author EQ 김민수
→ Pinecone filter: {'filter': {'author': {'$eq': '김민수'}}}
```

### 코드에서 정의하는 메서드

- `visit_comparison()`: `author EQ 김민수` 같은 조건 하나를 변환한다.
- `visit_operation()`: 여러 조건을 `$and` 또는 `$or`로 묶는다.
- `visit_structured_query()`: 의미 검색 문자열과 Pinecone 검색 옵션을 tuple로 반환한다.
- `accept(self)`: 조건 종류에 맞는 `visit_...()` 메서드를 자동으로 선택한다.


````python
from langchain_core.structured_query import (
    Comparator,
    Comparison,
    Operation,
    Operator,
    StructuredQuery,
    Visitor,
)

# Visitor : 조건 종류에 맞는 visit_... () 메서드를 호출할 수 있게 하는 기반 클래스
class PineconeFilterTranslator(Visitor):
    # 1. Pinecone filter에서 사용할 수 있는 비교·결합 연산을 제한한다.

    # allowed_comparators: 필드 값 하나를 비교하는 허용 연산 목록이다.
    # EQ·NE: 같음·다름, LT·LTE·GT·GTE: 크기 비교, IN·NIN: 목록 포함·미포함
    allowed_comparators = (
        Comparator.EQ,
        Comparator.NE,
        Comparator.LT,
        Comparator.LTE,
        Comparator.GT,
        Comparator.GTE,
        Comparator.IN,
        Comparator.NIN,
    )

    # allowed_operators: 여러 조건을 묶을 때 AND와 OR만 허용한다.
    allowed_operators = (Operator.AND, Operator.OR)

    # 2. LangChain 연산자 이름을 Pinecone 연산자 문자열로 바꾼다.

    # function: Comparator.EQ나 Operator.AND 같은 연산자 객체이다.
    # _validate_func(): function이 위의 허용 목록에 있는지 검사한다.
    # 반환 예시: Comparator.EQ → 'eq', Operator.OR -> 'or'
    def _format_function(self, function) -> str:
        self._validate_func(function)
        return f'${function.value}'

    # 3. metadata 조건 하나를 Pinecone 딕셔너리로 변환한다.

    # comparison 입력 예시: author EQ 김민수
    # comparison.attribute: metadata key인 'author'
    # comparison.value: 비교 값인 '김민수'
    # 반환 예시: {'author': {'$eq': '김민수'}}
    def visit_comparison(self, comparison: Comparison) -> dict:
        value = comparison.value

        # nin의 오른쪽 값은 목록이어야 하므로 단일 값도 list로 감싼다.
        # 변환 예시: '검색' → ['검색']
        if (
            comparison.comparator in (Comparator.IN, Comparator.NIN)
            and not isinstance(value, list)
        ):
            value = [value]

        # comparator도 _format_function()을 거쳐 EQ에서 '$eq'로 변환된다.
        return {
            comparison.attribute: {
                self._format_function(comparison.comparator): value
            }
        }

    # 4. 여러 metadata 조건을 and 또는 or 딕셔너리로 묶는다.

    # operation.arguments: Comparison 또는 중첩 Operation의 목록이다.
    # accept(self): 각 조건의 종류에 맞는 visit 메서드를 호출한다.
    def visit_operation(self, operation: Operation) -> dict:
        arguments = [
            argument.accept(self)
            for argument in operation.arguments
        ]

        # 반환 예시: {'$or': [첫 번째 filter, 두 번째 filter]}
        return {self._format_function(operation.operator): arguments}

    # 5. StructuredQuery 전체를 Vector Store가 사용할 두 값으로 나눈다.

    # 반환 tuple의 첫 값: embedding에 사용할 semantic query 문자열
    # 반환 tuple의 두 번째 값: Pinecone에 전달할 search_kwargs 딕셔너리
    def visit_structured_query(
        self,
        structured_query: StructuredQuery,
    ) -> tuple[str, dict]:
        # filter가 없으면 의미 검색만 실행하므로 빈 검색 옵션을 사용한다.
        if structured_query.filter is None:
            search_kwargs = {}
        else:
            # filter가 있으면 Comparison·Operation을 Pinecone 문법으로 변환한다.
            search_kwargs = {
                'filter': structured_query.filter.accept(self)
            }

        # 이 tuple은 뒤에서 SelfQueryRetriever가 Vector Store 검색에 사용한다.
        return structured_query.query, search_kwargs
````

## StructuredQuery 생성용 Chat Model 준비

`ChatOpenAI`는 사용자에게 답변하기 위해서가 아니라 자연어를 StructuredQuery로 변환하기 위해 사용한다. 객체 생성만으로는 API를 호출하지 않으며, 뒤에서 query constructor를 `invoke()`할 때 요청이 발생한다.

- `model`: 사용할 Chat Model 이름이다.
- `use_responses_api=True`: Responses API로 모델을 호출한다.
- `temperature=0`: query 구조의 불필요한 변화를 줄인다.
- `reasoning_effort='none'`: 별도 추론 토큰 없이 짧은 구조 변환을 수행한다.


````python
from langchain_openai import ChatOpenAI

llm=ChatOpenAI(
    model = OPENAI_LLM_MODEL, # gpt-5.6-luna
    use_responses_api=True,
    temperature=0,
    reasoning_effort='none'
)
````

## Self-query Retriever 구성 2: metadata 스키마와 객체 연결

`AttributeInfo` 두 개로 LLM이 사용할 metadata 스키마를 만든다. 그런 다음 LLM, Vector Store와 앞에서 정의한 translator를 `SelfQueryRetriever.from_llm()`으로 연결한다.

- `author`: 저자 한 명을 저장하므로 `string`이다.
- `category`: 문서 하나가 여러 범주를 가질 수 있으므로 `list[string]`이다.
- 기본 반환 개수: `search_kwargs={'k': 5}`를 사용한다.
- 수량 요청: `enable_limit=True`이므로 자연어 limit이 있으면 기본 k를 덮어쓴다.


````python
from langchain_classic.chains.query_constructor.schema import AttributeInfo
from langchain_classic.retrievers.self_query.base import SelfQueryRetriever

# metadata_field_info: LLM이 filter에 사용할 수 있는 metadata 필드 목록
metadata_field_info = [
    AttributeInfo(
        # name: Pinecone에 저장된 실제 metadata key이다.
        name='author',
        # #type: author 값 하나가 문자열임을 알린다.
        type='string',
        # description: 자연어의 저자 표현을 author 필드와 연결한다.
        description='문서를 작성한 저자 이름',
    ),
    AttributeInfo(
        name='category',
        type='list[string]',
        description='문서가 속한 카테고리 이름 목록',
    ),
]

# translator: StructuredQuery를 Pinecone 검색 인자로 바꾸는 객체
translator = PineconeFilterTranslator()

# llm: 자연어를 StructuredQuery로 바꾼다.
self_query_retriever = SelfQueryRetriever.from_llm(
    llm=llm,
    vectorstore=vector_store,
    document_contents='문서의 본문 내용',

    # LLM이 생성 가능한 필터의 필드를 author, category로 제한
    metadata_field_info=metadata_field_info,

    # StructuredQuery : LLM이 query를 분석해서 필터를 추출한 대답 양식
    # StructuredQuery를 Pinecone 필터 형식으로 번역
    structured_query_translator=translator,


    enable_limit=True,
    search_type='similarity',
    search_kwargs={'k': 5},
)
````

## 자연어 저자 조건으로 문서 검색

`invoke()`에 자연어 질문 하나를 전달하면 Self-query의 전체 흐름이 실행된다.

`자연어 질문 → StructuredQuery 생성 → Pinecone filter 변환 → vector 검색 → list[Document]`

- `Document`: 검색된 문서 하나를 나타내는 LangChain 객체이다.
- `page_content`: 검색된 본문 문자열이다.
- `metadata`: `doc_id`, `author`, `category` 같은 부가정보 딕셔너리이다.
- `list[Document]`: 관련성이 높은 순서로 반환된 Document 목록이며 LLM 답변을 생성하지 않는다.

실행 결과의 모든 `author`가 질문의 `김민수`와 일치하는지 확인한다.


````python
user_query = '김민수가 작성한 문서를 모두 찾아줘'

results = self_query_retriever.invoke(user_query)

for doc in results:
    print(doc.metadata['doc_id'], '/', doc.metadata['author'])
    print(doc.page_content)
    print()
````

## 자연어 질문이 Pinecone 검색 조건으로 바뀌는 과정

이번에는 Self-query Retriever의 내부 두 단계를 따로 확인한다. 이 셀에서는 Pinecone 검색을 실행하지 않는다.

1. `query_constructor`: 자연어를 `StructuredQuery(query, filter, limit)`으로 변환한다.
2. `translator`: StructuredQuery의 filter를 Pinecone filter 딕셔너리로 변환한다.

`김민수가 작성한 문서 2개`라는 질문은 다음 구조가 되는지 확인한다.

- `query`: 본문 주제가 없으므로 빈 문자열에 가까운 값이다.
- `filter`: `Comparison(Comparator.EQ, 'author', '김민수')` 형태이다.
- `limit`: `enable_limit=True`이므로 `2`가 될 수 있다.
- Pinecone 조건: `{'filter': {'author': {'$eq': '김민수'}}, 'k': 2}` 형태이다.

LLM 출력은 모델에 따라 표현이 달라질 수 있으므로 실제 필드값을 직접 확인한다.


````python
# query_constructor: Self-query Retriever 안에서 자연어를 StructuredQuery로 바꾸는 Runnable이다.
query_constructor = self_query_retriever.query_constructor

user_query = '김민수가 작성한 문서 2개만 찾아줘'
structured_query = query_constructor.invoke({'query': user_query})

# visit_structured_query(): StructuredQuery의 query와 filter를 Pinecone 검색 인자로 변환한다.
# 반환값: (검색 문자열, 검색 옵션 딕셔너리) tuple이다. 이 단계에서는 검색하지 않는다.
pinecone_query, pinecone_search_kwargs = translator.visit_structured_query(
    structured_query
)

# SelfQueryRetriever는 limit이 있으면 Pinecone의 반환 개수 k로 추가한다.
if structured_query.limit is not None:
    pinecone_search_kwargs['k'] = structured_query.limit

print('원본 질문:', user_query)
print('query:', repr(structured_query.query))
print('filter:', structured_query.filter)
print('limit:', structured_query.limit)
print('Pinecone query:', repr(pinecone_query))
print('Pinecone search kwargs:', pinecone_search_kwargs)
````

## OR로 연결된 카테고리 조건 확인

## 검색 결과 평가를 위한 qrels 변환

Self-query가 filter를 자동 생성하는 것만으로 검색 품질이 좋아졌다고 판단할 수는 없다. Dense와 Self-query의 검색 문서 ID를 질의별 정답표인 qrels와 비교해야 한다.

`검색 결과 문서 ID → qrels 정답과 비교 → 질의별 지표 → 전체 평균`

- qrels 입력: `D6=3;D14=2` 같은 문자열이다.
- 변환 결과: `{'D6': 3, 'D14': 2}` 같은 딕셔너리이다.
- key: 정답 문서 ID이다.
- value: 관련성 등급이며 이 실습에서는 1 이상을 정답으로 처리한다.

문자열을 딕셔너리로 바꾸면 검색된 `doc_id`가 정답 key에 포함되는지 정확히 검사할 수 있다.


````python
def parse_relevant(relevant_str: str) -> dict[str, int]:
    relevant_dict = {}

    # split(';'): 문서별 `ID=등급` 문자열을 목록으로 나눈다.
    for pair in relevant_str.split(';'):
        # split('='): 문서 ID와 문자열 등급을 분리한다.
        doc_id, grade_text = pair.split('=')
        grade = int(grade_text)

        # grade가 1 이상인 문서만 정답으로 보존한다.
        if grade > 0:
            relevant_dict[doc_id] = grade

    return relevant_dict
````

## 질의 하나의 검색 지표 계산

검색 결과는 관련 문서를 **찾았는지**뿐 아니라 **몇 위에서 찾았는지**도 평가해야 한다. `k=5`는 최대 반환 개수가 아니라 평가할 상위 순위의 범위인 cutoff이다. 결과가 5개보다 적으면 비어 있는 순위는 관련 없는 결과로 처리되어 P@5가 낮아진다.

- P@k(Precision at k): 상위 k개 중 정답 문서의 비율이다.
- R@k(Recall at k): 전체 정답 문서 중 상위 k개에서 찾은 비율이다.
- RR(Reciprocal Rank): 처음 등장한 정답 문서 순위의 역수이다. 1위이면 1.0, 2위이면 0.5이다.
- AP@k(Average Precision at k): 정답을 만난 각 순위의 Precision을 평균해 정답의 순서와 누락을 함께 반영한다.

함수는 네 값을 `(precision, recall, rr, ap)` tuple로 반환한다.


````python
def compute_metrics(
    predicted: list[str],
    relevant_dict: dict[str, int],
    k: int = 5,
) -> tuple[float, float, float, float]:
    # predicted[:k]: 검색 결과 중 평가할 상위 k개 문서이다.
    top_k = predicted[:k]

    # Boolean은 sum()에서 True=1, False=0으로 계산되어 정답 문서 수가 된다.
    hits = sum(doc_id in relevant_dict for doc_id in top_k)
    precision = hits / k

    # Recall은 찾은 정답 수를 qrels의 전체 정답 수로 나눈다.
    total_relevant = len(relevant_dict)
    recall = hits / total_relevant if total_relevant else 0.0

    # enumerate(start=1): 첫 검색 결과의 rank를 1부터 부여한다.
    # next(generator, 0.0): 첫 정답의 순위 역수를 반환하고 정답이 없으면 0.0을 반환한다.
    rr = next(
        (
            1 / rank
            for rank, doc_id in enumerate(predicted, start=1)
            if doc_id in relevant_dict
        ),
        0.0,
    )

    # 정답을 만난 순위마다 그 지점까지의 Precision을 누적한다.
    precision_sum = 0.0
    relevant_seen = 0
    for rank, doc_id in enumerate(top_k, start=1):
        if doc_id in relevant_dict:
            relevant_seen += 1
            precision_sum += relevant_seen / rank

    # min(total_relevant, k): 상위 k에서 찾을 수 있는 최대 정답 수를 분모로 사용한다.
    ap_denominator = min(total_relevant, k)
    ap = precision_sum / ap_denominator if ap_denominator else 0.0

    return precision, recall, rr, ap
````

## 모든 질의의 평균 검색 지표 계산

`evaluate_all()`은 Q01부터 Q30까지 같은 평가를 반복하고 질의별 지표의 평균을 계산한다.

- `method_results`: `query_id → 순서가 있는 문서 ID 목록` 딕셔너리이다.
- `queries_df`: 각 query_id의 자연어 질문과 qrels를 가진 DataFrame이다.
- `per_query_metrics`: 질의마다 `(P@k, R@k, RR, AP@k)`를 저장한 목록이다.
- `metric_array`: `(질의 수, 4)` shape의 NumPy 배열이다. 행은 질의, 열은 네 지표이다.
- 반환값: 전체 질의의 P@k·R@k·MRR·MAP 평균 딕셔너리이다.

MRR은 질의별 RR의 평균이고, MAP는 질의별 AP의 평균이다.


````python
import numpy as np

def evaluate_all(
    method_results: dict[str, list[str]],
    queries_df: pd.DataFrame,
    k: int = 5,
) -> dict[str, float]:
    per_query_metrics = []

    for _, row in queries_df.iterrows():
        relevant_dict = parse_relevant(row['relevant_doc_ids'])
        predicted = method_results[row['query_id']]
        per_query_metrics.append(
            compute_metrics(predicted, relevant_dict, k)
        )

    metric_array = np.asarray(per_query_metrics, dtype=float)

    # axis 0 방향으로 각 열의 mean을 계산해 전체 질의 평균을 만든다.
    return {
        'P@k': metric_array[:, 0].mean(),
        'R@k': metric_array[:, 1].mean(),
        'MRR': metric_array[:, 2].mean(),
        'MAP': metric_array[:, 3].mean(),
    }
````

## Dense 검색과 Self-query 검색 결과 수집

Self-query가 metadata 조건을 자동 생성하는 것만으로 검색 성능이 좋아졌다고 판단할 수는 없다. 같은 30개 질문을 두 방식으로 검색한 뒤, 반환된 문서 ID를 qrels 정답과 비교한다.

- Dense 기준선: metadata filter 없이 질문 전체를 embedding해 검색한다.
- Self-query: 질문을 의미 검색어와 metadata 조건으로 분리해 검색한다.
- 수집 결과: `query_id → 검색된 doc_id 목록`으로 저장한다.

두 결과의 key와 반환 개수를 같게 만들어 앞에서 정의한 `evaluate_all()`에 전달한다.


````python
from tqdm import tqdm

dense_results = {}
self_query_results = {}

# tqdm(iterable, total): 반복 진행률을 표시하며 total=30이 전체 작업 수가 된다.
# iterrows(): 각 반복에서 사용하지 않는 행 번호 `_`와 질의 한 행 `row`를 반환한다.
for _, row in tqdm(queries_df.iterrows(), total=len(queries_df)):
    qid = row['query_id']
    query_text = row['query_text']

    # similarity_search(query, k): 질문 전체를 embedding해 상위 5개 Document를 반환한다.
    # 이 경로에는 자동 metadata filter가 없다.
    dense_docs = vector_store.similarity_search(
        query=query_text,
        k=5,
    )

    # list comprehension: Document 목록에서 평가에 필요한 doc_id만 같은 순서로 추출한다.
    dense_results[qid] = [
        doc.metadata['doc_id']
        for doc in dense_docs
    ]

    # invoke(query): LLM이 만든 query·filter·limit으로 Pinecone을 검색한다.
    self_query_docs = self_query_retriever.invoke(query_text)
    self_query_results[qid] = [
        doc.metadata['doc_id']
        for doc in self_query_docs
    ]
````

두 검색 결과를 같은 qrels와 `k=5`로 평가해 비교표를 만든다.

- `Dense`: metadata filter 없이 질문 전체를 embedding한 기준선이다.
