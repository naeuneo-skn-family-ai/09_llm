# LangChain

## LangChain이란

[LangChain](https://docs.langchain.com/oss/python/langchain/overview)은 모델, 프롬프트, 검색기, 도구처럼 LLM 애플리케이션에 **필요한 기능을 일관된 방식으로 연결**하는 프레임워크이다.

OpenAI, Anthropic, Google 등 공급자가 달라도 비슷한 인터페이스로 모델을 호출할 수 있고, 한 단계의 출력을 다음 단계의 입력으로 전달하는 실행 흐름을 구성할 수 있다.

LangChain이 모델의 답변을 직접 보유하거나 학습시키는 것은 아니다.

모델이 문장을 생성하면 LangChain은 그 모델을 어떤 프롬프트로 호출할지, 어떤 외부 문서를 참고할지, 어떤 도구를 사용할지와 같은 애플리케이션 흐름을 관리한다.


## 등장 배경

LangChain은 Harrison Chase가 2022년에 시작한 오픈소스 프로젝트이다.
초기에는 GPT 계열 모델을 프롬프트, 외부 데이터, 도구와 연결할 때 반복해서 작성하던 코드를 줄이는 데 초점을 두었다.

LLM API만 사용해도 텍스트를 생성할 수 있지만, 실제 애플리케이션에는 모델 호출 이외의 작업이 필요하다.

- 모델 공급자마다 다른 호출 방식과 응답 형식을 처리해야 한다.
- 프롬프트에 사용자 입력과 외부 데이터를 일정한 형식으로 결합해야 한다.
- 문서 검색, 결과 파싱, 도구 실행 같은 단계를 순서대로 연결해야 한다.
- 실행 단계가 많아지면 어느 단계에서 오류가 발생했는지 추적하기 어려워진다.

## LangChain 생태계의 역할

- **LangChain**: 모델, 프롬프트, 검색기, 도구를 조합해 LLM 애플리케이션과 에이전트를 만든다.
- **LangGraph**: 분기, 반복, 상태 보존, 사람 승인처럼 세밀한 제어가 필요한 실행 흐름을 그래프로 구성한다.
- **LangSmith**: 실행 과정을 trace로 기록하고 디버깅, 평가, 모니터링한다.


LangChain은 이러한 기능을 **재사용 가능한 구성 요소와 공통 실행 방식**으로 제공한다.

현재는 모델과 도구를 반복 호출하는 에이전트 개발을 중심으로 발전했으며, 복잡한 상태와 실행 제어는 LangGraph, 실행 관찰과 평가는 LangSmith가 담당한다.



모델을 한 번 호출하는 간단한 프로그램은 OpenAI SDK만으로도 충분하다.
단, 여러 모델을 교체하거나 검색과 도구를 연결하고, 반복해서 사용할 실행 흐름을 구성할 때 LangChain의 장점이 커진다.

## LangChain 핵심 구성 요소와 주요 패키지

### 핵심 구성 요소

LangChain 애플리케이션은 필요한 구성 요소를 선택하고 연결해 만든다. 모든 애플리케이션이 아래 요소를 전부 사용하는 것은 아니며, 목적에 따라 모델 호출만 사용하거나 검색과 도구를 함께 사용할 수 있다.

<img src="https://d.pr/i/WA9QHw+" alt="LangChain이 Models, Prompts, Chains, Embedding and Vector Stores, Agents를 연결하는 구성 요소 지도" width="900"/>

Models, Prompts, Chains, Embedding and Vector Stores, Agents가 독립된 기능이면서 서로 연결될 수 있음을 보여 준다.
화살표는 가능한 연결 관계를 나타내며 고정된 실행 순서를 뜻하지 않는다.

- **Model I/O**: Prompt가 모델에 전달할 입력을 만들고, Model이 응답을 생성하며, Output Parser가 응답을 후속 코드에서 사용할 형식으로 바꾼다.
- **Retrieval**: 문서를 분할하고 임베딩한 뒤 Vector Store에서 관련 내용을 검색해 모델에 근거를 제공한다.
- **Chains와 Runnable**: 여러 처리 단계를 연결하고, 앞 단계의 출력을 다음 단계의 입력으로 전달한다.
- **Memory와 상태**: 대화 이력과 작업 진행 상태를 다음 모델 호출이나 다음 실행 단계에서 다시 사용할 수 있게 보존한다. 복잡한 상태 관리는 LangGraph가 담당한다.
- **Agents와 Tools**: 모델이 문제를 해결하는 데 필요한 도구를 선택하고, 도구 실행 결과를 이용해 다음 행동이나 최종 답변을 결정한다.


### 주요 패키지

LangChain은 모든 기능을 하나의 거대한 패키지에 넣지 않고 역할별 패키지로 나눈다. 필요한 기능만 설치할 수 있고, 모델 공급자별 패키지를 교체하기도 쉽다.

- `langchain`: 에이전트 생성, 통합 모델 초기화처럼 애플리케이션을 구성하는 상위 인터페이스를 제공한다.
- `langchain-core`: message, prompt, output parser, Runnable처럼 공급자와 관계없이 사용하는 핵심 추상화를 제공한다. 일반적으로 `langchain`을 설치하면 호환 버전이 함께 설치된다.
- `langchain-openai`, `langchain-anthropic`: OpenAI와 Anthropic처럼 특정 모델 공급자를 연결하는 통합 패키지이다.
- `langchain-text-splitters`: 긴 문서를 검색과 임베딩에 적합한 작은 단위로 분할한다.
- `langchain-community`: 여러 서드파티 통합을 모아 제공하던 패키지이다. 기존 코드에서는 사용할 수 있지만 새 코드에서는 유지보수 중인 공급자별 통합 패키지나 직접 구현을 먼저 검토한다.
- `langchain-classic`: 이전 버전의 미리 구성된 chain, retriever, indexing 기능과 호환성을 제공한다. 새 수업의 기본 경로로 사용하지 않는다.

LangChain v1에서는 `langchain` 패키지가 모델과 에이전트 중심으로 단순해졌고, 이전 chain과 retriever 일부가 `langchain-classic`으로 이동했다. 오래된 예제와 import 경로가 다른 이유를 이해하는 데 필요한 변화이므로 이 차이만 기억하면 된다. 자세한 변경 사항은 [LangChain v1 마이그레이션 가이드](https://docs.langchain.com/oss/python/migrate/langchain-v1)에서 확인한다.
